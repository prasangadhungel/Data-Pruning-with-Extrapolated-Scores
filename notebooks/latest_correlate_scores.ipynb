```python
def get_top_keys(dictionary, percentage=0.9):
    sorted_keys = sorted(dictionary, key=dictionary.get, reverse=True)  # Sort keys by value in descending order
    top_n = int(len(sorted_keys) * percentage)  # Get the top percentage
    return set(sorted_keys[:top_n])  # Return as a set for easy intersection
```
