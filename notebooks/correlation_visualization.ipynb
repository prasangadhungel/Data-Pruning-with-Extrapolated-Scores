```python
import seaborn as sns
import matplotlib.pyplot as plt
import json
import numpy as np
```

```python
with open('/nfs/homedirs/dhp/unsupervised-data-pruning/scores/prune/CIFAR10_dynamic_uncertainty_0.json') as f:
    dynamic_uncertainty_score_superset_dict = json.load(f)
```

```python
with open('/nfs/homedirs/dhp/unsupervised-data-pruning/scores/extrapolation/extrapolated/gnn_extrapolation_CIFAR10_resnet50-self-trained_k_10_seed_20000_euclidean.json') as f:
    dynamic_uncertainty_extrapolated = json.load(f)
```

```python
original = np.array([dynamic_uncertainty_score_superset_dict[str(i)] for i in range(len(dynamic_uncertainty_extrapolated.keys())) if dynamic_uncertainty_score_superset_dict[str(i)] != dynamic_uncertainty_extrapolated[str(i)]])
extrapolated = np.array([dynamic_uncertainty_extrapolated[str(i)] for i in range(len(dynamic_uncertainty_extrapolated.keys())) if dynamic_uncertainty_score_superset_dict[str(i)] != dynamic_uncertainty_extrapolated[str(i)]])
```

```python
original.shape, extrapolated.shape
```

```python
plt.figure(figsize=(8, 6))  # Adjust figure size if needed
plt.scatter(original, extrapolated, marker='o', alpha=0.5, edgecolors='black') # Use scatter plot
plt.xlabel('Original Scores')
plt.ylabel('KNN Extrapolated (k=10)')
plt.title('Places 365 Scores')
plt.grid(True, linestyle='--', alpha=0.6) # Add a grid for better readability
plt.tight_layout() # Adjust layout for better spacing
plt.show()
```

```python
plt.hist(original, bins=20)
plt.title("Original Scores")
plt.show()
```

```python
plt.hist(extrapolated, bins=20)
plt.title("Extrapolated Scores (K=10)")
plt.show()
```
